```python
import numpy as np
```

```python
nombres = []
nombres.append('Andres')
nombres.append('Diana')
nombres.append('Laura')

apellidos = []
apellidos.append('Trujillo')
apellidos.append('Henao')
apellidos.append('Rodriguez')

edad = []
edad.append(32)
edad.append(43)
edad.append(53)

cedula = []
cedula.append(123)
cedula.append(456)
cedula.append(324)

genero = ['M', 'F', 'F']


# La base conecta un diccionario con otro diccionario
usuarios = {}
n = len(cedula)

for i in range(n):
  usuarios[cedula[i]] = {
      'nombre': nombres[i],
      'apellido': apellidos[i],
      'edad': edad [i],
      'genero': genero[i]
  }

### Consultas
usuarios[324]['edad'] # Cual es la edad de Laura?

### Cambios
usuarios[456]['apellido'] = 'Bastidas' # Como modificar el apellido?
usuarios[456]
```

```python
# La visualizacion
import seaborn as sns

### visualizar y: edad, x: usuario
# 1. Extraer la informacion de la base de datos

edades = [ persona['edad'] for persona in usuarios.values() ] # for loop por compresion
nombres = [ persona['nombre'] for persona in usuarios.values() ]

# 2. Visualizacion de los datos - Categorical plot
sns.catplot(x=nombres, y=edades, kind='bar')
```

```python
### Cómo podemos añadir la información del genero
# 1. Primer filtro donde escogemos a los hombres
edades = [ persona['edad'] for persona in usuarios.values() if persona['genero'] == 'F' ]
nombres = [ persona['nombre'] for persona in usuarios.values() if persona['genero'] == 'F' ]

# 2. Visualizacion de los datos - Categorical plot
sns.catplot(x=nombres, y=edades, kind='bar')
```

```python
### Nombres vs genero
nombres = [ persona['nombre'] for persona in usuarios.values() ]
generos = [ persona['genero'] for persona in usuarios.values() ]

sns.stripplot(x=nombres, y=generos, size=10)
```
