```python
# Concatenar los 10 archivos en uno solo
lista_archivos = []

for i in range(1, 11):
  ruta_carpeta = ('Proceso_Header_{}/*.csv'.format(i)) # Ruta carpeta por carpeta
  archivos_carpeta = glob.glob(ruta_carpeta) # Buscar los archivos con patron especifico
  lista_archivos.extend(archivos_carpeta)

df_concat_original = pd.concat(
    [pd.read_csv(f) for f in lista_archivos], ignore_index=True
)

ruta_proyecto = '/content/drive/MyDrive/EDCO_IntroCienciaDatos/Proyectos'
os.chdir(ruta_proyecto)
df_concat_original.to_csv('archivo_concatenado.csv', index=False, header=True)
```

```python

# Problema con los datos, paciente real que tiene ambas fechas 'muerte y recuperación'
## Fecha de recuperación y tipo (82 nulos) -> verificar Probablemente coinciden con las personas que fallecieron (83 casos)
#  Verificar si esos nulos corresponden a los fallecidos. Si es así, puedes rellenar Tipo de recuperación con 'No aplica (Fallecido)'

# Creamos columnas temporales que indiquen True si es nulo y False si no lo es
fallecido = df_concat['Fecha de muerte'].isna() # False cuando NO esta vacio
sin_recuperacion = df_concat['Fecha de recuperación'].isna() # False cuando NO esta vacio

# Generamos una tabla cruzada - No tienen fecha de recup, personas que fallecieron
print(pd.crosstab(fallecido, sin_recuperacion, rownames=['Muerte es Nula'], colnames=['Recuperación es Nula']))

# Encontrar al paciente que tiene AMBAS fechas registradas
paciente_error = df_concat[df_concat['Fecha de muerte'].notna() & df_concat['Fecha de recuperación'].notna()]
print(paciente_error[['ID de caso', 'Fecha de muerte', 'Fecha de recuperación', 'Estado']])


###################
paciente_error = df_concat[df_concat['Fecha de muerte'].notna() & df_concat['Fecha de recuperación'].notna()]

print("El paciente con la anomalía real es:")
print(paciente_error[['ID de caso', 'Fecha de muerte', 'Fecha de recuperación', 'Estado']]) # Hay dos pacientes asi

# Ver la columna 'Recuperado' para desempatar
print("Ver columna Recuperado:")
print(df_concat.loc[[1815, 1852], ['ID de caso', 'Fecha de muerte', 'Fecha de recuperación', 'Recuperado']]) # Fecha de muerte son errores

#-------------------------------------------
# Borrar fecha de muerte que está errada y asignar Estado = 'Leve'
indices_anomalia = [1815, 1852] # Índices de los pacientes con la anomalía

df_concat.loc[indices_anomalia, 'Fecha de muerte'] = np.nan # Borrar la fecha de muerte falsa poniéndole NaN

df_concat.loc[indices_anomalia, 'Estado'] = 'Leve' # Corregir su Estado a 'Leve' (en lugar del NaN que tenían)

# Verificar
print("Registros corregidos exitosamente:")
print(df_concat.loc[indices_anomalia, ['ID de caso', 'Fecha de muerte', 'Fecha de recuperación', 'Estado']])
```

```python
# Valores nulos/faltantes - Eliminar, imputo, dejarlos en el documento
## Fecha de muerte (Solo 83 no-nulos de 3000): Significa que el paciente sobrevivió
df_concat['Fecha de muerte'] = df_concat['Fecha de muerte'].fillna('No aplica')

## Nombre del grupo étnico (Solo 32 no-nulos): No pertenece a un grupo étnico específico
df_concat['Nombre del grupo étnico'] = df_concat['Nombre del grupo étnico'].fillna('No aplica')

## Fecha de inicio de síntomas: Rellenar con la 'Fecha de diagnóstico' quizás asintomaticos
df_concat['Fecha de inicio de síntomas'] = df_concat[
    'Fecha de inicio de síntomas'
    ].fillna(df_concat['Fecha de diagnóstico'])

###################

## Estado y condición del paciente: Para no sesgar análisis, se convierten en nueva categoría "Desconocido"
df_concat['Estado'] = df_concat['Estado'].fillna('Desconocido')
df_concat['Ubicación del caso'] = df_concat['Ubicación del caso'].fillna('Desconocido')

###################

# Revisar el estado de la tabla actualmente
df_concat.info()
```

```python
## Limpieza de nulos residuales detectados en el .info()
# Recuperado (Faltaban 20): Rellenar basado en si tiene fecha de muerte o no
df_concat['Recuperado'] = df_concat['Recuperado'].fillna(
    df_concat['Fecha de muerte'].apply(lambda x: 'Fallecido' if x != 'No aplica' else 'Recuperado')
)

# Fecha de diagnóstico (Faltaba 1): Rellenar con la fecha de notificación
df_concat['Fecha de diagnóstico'] = df_concat['Fecha de diagnóstico'].fillna(df_concat['Fecha de notificación'])

# Pertenencia étnica (Faltaba 1): Rellenar con la moda de la columna
df_concat['Pertenencia étnica'] = df_concat['Pertenencia étnica'].fillna(df_concat['Pertenencia étnica'].mode()[0])

###################

# Agregamos un fillna general por si algún vivo quedó sin estos datos
df_concat['Fecha de recuperación'] = df_concat['Fecha de recuperación'].fillna('No aplica')
df_concat['Tipo de recuperación'] = df_concat['Tipo de recuperación'].fillna('No aplica')

# Revisar el estado de la tabla actualmente
df_concat.info()
```
