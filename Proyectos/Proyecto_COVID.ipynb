```python
# Librerias y sistema operativo
import os

import glob
import pandas as pd
```

```python
# Movernos a la carpeta base con los archivos/subcarpetas
ruta = '/content/drive/MyDrive/EDCO_IntroCienciaDatos/Semana2/BaseDatos'
os.chdir(ruta)
!ls # Listar contenido
```

```python
# Concatenar los 10 archivos en uno solo
lista_archivos = []

for i in range(1, 11):
  ruta_carpeta = ('Proceso_Header_{}/*.csv'.format(i)) # Ruta carpeta por carpeta
  archivos_carpeta = glob.glob(ruta_carpeta) # Buscar los archivos con patron especifico
  lista_archivos.extend(archivos_carpeta)

df_concat = pd.concat(
    [pd.read_csv(f) for f in lista_archivos], ignore_index=True
)

ruta_proyecto = '/content/drive/MyDrive/EDCO_IntroCienciaDatos/Proyectos'
os.chdir(ruta_proyecto)
df_concat.to_csv('archivo_concatenado.csv', index=False, header=True)
```
