Revisar nuestras bases de datos

```python
BaseDatos = {
    1001: { 'Nombre': 'Pepito', 'Apellido': 'Perezito', 'Edad': 35, 'Genero': 'M' },
    1002: { 'Nombre': 'Pepita', 'Apellido': 'Perezita', 'Edad': 25, 'Genero': 'F' },
    1003: { 'Nombre': 'Pepite', 'Apellido': 'Perezite', 'Edad': 28, 'Genero': 'M' },
}
```

```python
# Primer dataframe
df = pd.DataFrame.from_dict(BaseDatos, orient='index')
df
```

```python
# Elegir una columna de los datos
df['Edad']
```

```python
# Pandas básico
df.index.name = 'ID'
df
```

```python
# Podemos convertir el ID en una columna
df = df.reset_index()
df
```

```python
# Queremos escoger solo unas columnas especificas
df1 = df[['Edad', 'Genero']] # Lista con las columnas que quiero
df1
```

```python
# Aplicar mascaras "masking" o filtros
# Personas mayores a 25
mask = df['Edad'] > 25 # Esto es una condición
df2 = df[mask]
df2
```

```python
mask2 = df[ 'Genero' ] == 'M'
mask2

df[mask2]
```

```python
# Filtro 1 es que tenga más de 25 y el dos que sea del genero másculino
df[mask & mask2] # Y
df[mask | mask2] # O
```

```python
# Cómo consultar una entrada en particular
df.loc[1] # Ubicación 1
```

```python
# Tomar una porción del dato, estilo álgebra lineal - Slicing - Solo columnas contiguas
# La fila 0, la fila 1... necesitamos las filas y las columnas o features
df.iloc[1,:] # Todos
df.iloc[1,1:4] # Solo el nombre y la edad

df[['Nombre', 'Genero']].iloc[:2,:] # Elijo las columnas que quiero y entonces tomo la opción
```
