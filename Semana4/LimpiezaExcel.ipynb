```python
Database = pd.read_excel('Datos/Alcaceres.xlsx')
Database
```

```python
# Que deberia hacer con los NAN de los nombre y apelldios
Database = Database.fillna("")
Database
```

```python
Database.keys()
```

```python
NombreCompleto = Database['Apellido 1'] + " " + Database['Apellido 2'] + " " + Database['Nombre 1'] + " " + Database['Nombre 2']
NombreCompleto
```

```python
# Asignamos el nombre completo en algun lugar
Database.insert( loc=10, column='Nombre', value=NombreCompleto )
```

```python
Database.iloc[:,:11]
```

```python
# Cambiar le nombre de una columna
Database = Database.rename( columns={ 'Localidad (NOMBRE)' : 'Localidad' } )
```

```python
Database.keys()
```

```python
Features = ['Tipo de Documento', \
            'Número de documento', \
            'Nombre', \
            'Número de contacto 1', \
            'A que EPS se encuentra afiliado', \
            'Localidad', \
            'Fecha sesión ', \
            'Personas sesión ']
Features
```

```python
for i in range(1,9):
  print(i)

  Columns = Features.copy()

  Columns[-1] += str(i)
  Columns[-2] += str(i)

  #print(Columns)
  basemask = Database[Columns]

  #Filtrar por sesion
  mask = basemask[ Columns[-1] ] == 1

  basemask = basemask[mask]

  # Podemos escribir
  output = pd.ExcelWriter( 'Alcaceres_sesion_'+str(i)+'.xlsx', engine='xlsxwriter' )
  basemask.to_excel(output, sheet_name='Alcaceres', index=False)
  output.close()
```

```python
# Abrimos el excel una sola vez
output = pd.ExcelWriter( 'Alcaceres_consolidad.xlsx', engine='xlsxwriter' )

for i in range(1,9):
  print(i)

  Columns = Features.copy()

  Columns[-1] += str(i)
  Columns[-2] += str(i)

  #print(Columns)
  basemask = Database[Columns]

  #Filtrar por sesion
  mask = basemask[ Columns[-1] ] == 1

  basemask = basemask[mask]

  # Lo que si cambia es el sheet-name
  basemask.to_excel( output, sheet_name='Sesion_'+str(i), index=False )

output.close()
```
